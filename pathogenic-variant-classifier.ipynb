# Data Preprocessing and EDA

```python
missing_counts = df.isnull().sum() #missing value for each column
print(missing_counts[missing_counts > 0])
```

```python
threshold = len(df) * 0.5
df = df.dropna(thresh = threshold, axis = 1) #dropping cols where 50% of rows are missing
```

```python
num_cols = df.select_dtypes(include = ['int64', 'float64']).columns #identifying numerical columns

for col in num_cols:
    if df[col].isnull().sum() > 0:
        df[col] = df[col].fillna(df[col].median()) #filling missing numerical vals with median
```

```python
cat_cols = df.select_dtypes(include = ['str', 'object', 'category']).columns #identifying categorical text columns

for col in cat_cols:
    if df[col].isnull().sum() > 0:
        df[col] = df[col].fillna('Unknown') #filling missing text vals with 'Unknown'
```

```python
features_to_encode = [col for col in cat_cols if col != 'CLASS'] #replace 'CLASS' with target column name if different

ohe = OneHotEncoder(drop='first', sparse_output=True, handle_unknown='ignore')
encoded_sparse = ohe.fit_transform(df[features_to_encode])
df_numeric = df.drop(columns=features_to_encode) #drop original text columns from dataframe
X_sparse = sp.hstack([df_numeric.values, encoded_sparse], format='csr') #combining numeric data with sparse encoded matrix
```
